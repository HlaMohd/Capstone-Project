```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost",
        "Neural Network"
    ],
    "Attack Precision": [0.43, 0.98, 0.94, 0.99],
    "Attack Recall": [0.82, 1.00, 1.00, 0.98],
    "Attack F1": [0.57, 0.99, 0.97, 0.98],
    "Accuracy": [0.93, 1.00, 1.00, 1.00]
})

results
```

```python
results.sort_values(
    by="Attack F1",
    ascending=False
)
```

```python
results.plot(
    x="Model",
    y=["Attack Precision", "Attack Recall", "Attack F1"],
    kind="bar",
    figsize=(10, 6)
)

plt.title("Model Performance Comparison")
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.savefig("../images/model_comparison.png")
plt.show()

```
